```python
import numpy as np
#hinh thang
def trapezoid(f, a, b, n, M2): 
  if n<=0:
        print('Số khoảng chia phải lớn hơn 0')
        return None
  sum=0
  h=(b-a)/n #dòng 1 thuật toán 14
  for i in range(1,n):
      sum+=f(a+i*h)
  I=(h/2)*(f(a)+2*sum+f(b)) #dòng 5 thuật toán 14
  E=M2*(b-a)**3/(12*n**2)
  return I, E

#hinh chu nhat
def midpoint(f, a, b, n, M2) -> tuple[float, float]: #(I, E)
    if not callable(f):
            raise(Exception('tham so f phai la mot ham so hop le'))
        if a >= b:
            raise(Exception('can duoi a phai nho hon can tren a'))
        if not isinstance(n, int):
            raise(Exception('so khoang chia n phai la so nguyen duong'))
        if M2<0:
            raise(Exception('can sai so M2 phai la so ko am'))
        h = (b-a)/n #dòng 1- thuật toán 13
        Sum_val = 0.0
        for i in range(n): #dòng 2 - thuật toán 13
            midpoint_val = ((a+i*h) + (a + (i+1)*h))/2 #dòng 3 - thuật toán 13
            Sum_val += f(midpoint_val) #dòng 4 - thuật toán 13
        I_rec = h*Sum_val #dòng 6 - thuật toán 13
        E = (M2 * (b-a) * (h**2))/24
        return I_rec, E #dòng7 - thuật toán 13
def simpson_1_3(f, a, b, n):
    """Tính tích phân xác định bằng phương pháp Simpson 1/3.

    Tên thuật toán: Công thức Simpson 1/3 tổng quát (Composite Simpson s 1/3
    Rule)

    Require (Đầu vào):
        f : Hàm số cần tính tích phân (callable).
        a : Cận dưới của tích phân (float).
        b : Cận trên của tích phân (float).
        n : Số khoảng chia (int, phải là số chẵn, n >= 2).

    Ensure (Đầu ra):
        Giá trị xấp xỉ của tích phân (float).
    """
    # --- Bước kiểm tra điều kiện biên ---
    if n % 2 != 0:
        raise ValueError(
            f"LỖI: Số khoảng chia n = {n} phải là một SỐ CHẴN cho phương pháp Simpson 1/3!"
        )
    if a >= b:
        raise ValueError("LỖI: Cận dưới a phải nhỏ hơn cận trên b!")

    # --- Bắt đầu thuật toán ---
    # dòng 1 - Tính bước nhảy h
    h = (b - a) / n

    # dòng 2 - Khởi tạo tổng bằng giá trị hai đầu mút
    tong = f(a) + f(b)

    # dòng 3 - Vòng lặp tính tổng các điểm có chỉ số chẵn (hệ số 2)
    for i in range(2, n - 1, 2):
        # dòng 4 - Cộng dồn các giá trị tại nút chẵn
        tong += 2 * f(a + i * h)

    # dòng 5 - Vòng lặp tính tổng các điểm có chỉ số lẻ (hệ số 4)
    for i in range(1, n, 2):
        # dòng 6 - Cộng dồn các giá trị tại nút lẻ
        tong += 4 * f(a + i * h)

    # dòng 9 - Nhân với h/3
    ket_qua = tong * (h / 3)
    # dòng 10 - Kết quả
    return ket_qua
#simpson 3/8
    if not isinstance(n, (int, np.integer)) or isinstance(n, (bool, np.bool_)):
        raise TypeError("n phai la so nguyen!")

    if n <= 0 or n % 3 != 0:
        raise ValueError("n phai la so duong chia het cho 3!")

    if a >= b:
        raise ValueError("Can a phai nho hon can b!")

    # Buoc 2: Tinh buoc nhay
    h = (b - a) / n

    # Buoc 3: Khoi tao tong hai dau mut
    tong = f(a) + f(b)

    # Buoc 4: Tinh tong cac diem ben trong
    for i in range(1, n):
        if i % 3 == 0:
            tong += 2 * f(a + i * h)
        else:
            tong += 3 * f(a + i * h)

    # Buoc 5: Tinh ket qua
    return tong * 3 * h / 8
```
