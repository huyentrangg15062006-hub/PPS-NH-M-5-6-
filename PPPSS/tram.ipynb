```python
import numpy as np
"""
    Tinh tich phan xac dinh bang phuong phap Simpson 3/8.

    Require (Dau vao):
        f : Ham so can tinh tich phan.
        a : Can duoi cua tich phan.
        b : Can tren cua tich phan.
        n : So khoang chia (so nguyen duong, chia het cho 3).

    Ensure (Dau ra):
        Gia tri xap xi cua tich phan.
"""
    # Buoc 1: Kiem tra dieu kien
    if not isinstance(n, (int, np.integer)) or isinstance(n, (bool, np.bool_)):
        raise TypeError("n phai la so nguyen!")

    if n <= 0 or n % 3 != 0:
        raise ValueError("n phai la so duong chia het cho 3!")

    if a >= b:
        raise ValueError("Can a phai nho hon can b!")

    # Buoc 2: Tinh buoc nhay
    h = (b - a) / n

    # Buoc 3: Khoi tao tong hai dau mut
    tong = f(a) + f(b)

    # Buoc 4: Tinh tong cac diem ben trong
    for i in range(1, n):
        if i % 3 == 0:
            tong += 2 * f(a + i * h)
        else:
            tong += 3 * f(a + i * h)

    # Buoc 5: Tinh ket qua
    return tong * 3 * h / 8
```
