```python
#thuật toán 14: Phương pháp hình thang
'''
Require:
    -f: hàm số cần tính tích phân
    -a: cận dưới
    -b: cận trên
    -n: số khoảng chia (n>0)
    -M2: chặn trên đạo hàm cấp 2
    -h: độ rộng khoảng chia
Ensure:
    -trả về giá trị tuple[I, E]: giá trị tích phân xấp xỉ I và cận sai số E
'''
def trapezoid(f, a, b, n, M2): 
  if n<=0:
        print('Số khoảng chia phải lớn hơn 0')
        return None
  sum=0
  h=(b-a)/n #dòng 1 thuật toán 14
  for i in range(1,n):
      sum+=f(a+i*h)
  I=(h/2)*(f(a)+2*sum+f(b)) #dòng 5 thuật toán 14
  E=M2*(b-a)**3/(12*n**2)
  return I, E
'''
Require:
    -x: danh sách mảng dữ liệu x
    -y: danh sách mảng dữ lệu y
    -h: độ rộng khoảng chia
Ensure:
    -trả về giá trị area: tổng diện tích bề mặt cắt ngang xấp xỉ tính được 
'''
def trapezoid_table(x, y):
    area=0
    for i in range(len(x)-1):
        h=x[i+1]-x[i]
        area+=0.5*h*(y[i]+y[i+1])
    return area
#tính gần đúng tích phân (5.11 phiếu bài tập)
def f(x):
    return x**5+3*x**3-2
print('Giá trị gần đúng và cận sai số của tích phân là:',trapezoid(f, 0, 2, 10, 10**-3))

#tính lưu lượng dòng chảy qua mặt cắt sông (ví dụ lấy trong báo cáo cô gửi)
x=[0, 4, 8, 12, 16, 20, 24, 28, 32, 36, 40]
y=[0, 1.2, 2.5, 3.6, 4.4, 4.8, 4.6, 3.9, 2.8, 1.5, 0]
v=0.8 #vận tốc dòng chảy trb
dien_tich=trapezoid_table(x,y)
luu_luong=dien_tich*v
print('diện tích mặt cắt ngang là:', dien_tich)
print('Lưu lượng dòng chảy là:', luu_luong)
    
```
